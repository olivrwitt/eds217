```python
### The filter pattern

# df[df['column'] <comparison> value] 
```

```python
high_readings = goleta[goleta['value'] > 20]
high_readings.shape
```

```python
print('before:', goleta.shape)
print('after: ', high_readings.shape)
```

```python
goleta['parameter'].value_counts()
```

### two conditions at once

```python
valid = (goleta['parameter'] == 'pm25') & (goleta['value'] > 9)
goleta[valid].describe()
```

### Turning a condition around with ~
this is used for negation in pandas

```python
not_pm25 = ~(goleta['parameter'] == 'pm25')
goleta[not_pm25].describe()
```

### Matching a list of values with .isin()

```python
particulates = goleta[goleta['parameter'].isin(['pm25', 'pm10'])]
particulates.shape
```

```python
of_interest = ['pm25', 'pm10']
goleta[goleta['parameter'].isin(of_interest)].shape
```
