```python
import pandas as pd
```

```python
url = 'https://github.com/TheEconomist/banana-index-data/releases/download/1.0/bananaindex.csv'
```

```python
banana = pd.read_csv(url, index_col = 'entity')
```

```python
banana = banana.drop(['year', 'Banana values', 'type', 'Chart?', 'Unnamed: 16'], axis='columns')
```

```python
banana.loc['Bananas', 'Bananas index (kg)']
```

```python
banana.sort_values('Bananas index (kg)', ascending=False).head(10)
```

```python
banana.sort_values('Bananas index (1000 kcalories)', ascending=False).head(10)
```

```python
banana.sort_values('Bananas index (100g protein)', ascending=False).head(10)
```

These lists contain many of the same foods, but differ in key ways. When the denominator is a kilogram, all of the top entries are meats. When the denominator changes from a kilogram to a thousand calories, some fruits enter the top (higher density/calorie content).  When looking at 100g protein, sugar rises to the top due likely to its very low protein content.  The only food that appears in all three is beef.
