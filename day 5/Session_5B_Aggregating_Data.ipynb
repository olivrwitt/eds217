```python
import pandas as pd

url = 'https://eds-217-essential-python.github.io/data/national_parks.csv'
parks = pd.read_csv(url)

# Remove the rows that have park Totals in them and make a new dataframe:
by_year = parks[~(parks['year'] == 'Total')]
national_parks = by_year[by_year['unit_type'] == 'National Park'].copy()

national_parks.shape
```

```python
national_parks.groupby('region')['visitors'].mean()
```

```python
national_parks.groupby('region')['visitors'].count()
```

```python
# .agg() removes the eye-matching by allowing you to do more than one calculation at a time
# Give it a list of aggregation names and it runs all of them on the same grouping and 
# puts the answers side by side:

national_parks.groupby('region')['visitors'].agg(['count', 'mean', 'max'])

# the result is a data frame rather than a series
```

```python
# Group national_parks by state and use .agg() with a list to get the count, mean and max of visitors. 
# Find a state whose mean looks impressive, but which is based on a limited sample number.

national_parks.groupby('state')['visitors'].agg(['count', 'mean', 'max'])

# Ohio has an impressively high mean, but based only on 39 samples
```

```python
# Several columns, summarised differently:

# for example: take the mean of this column, the sum of that one, 
# the number of distinct values in a third, etc.

# we do not select a single column, instead we run .agg() against the groupby object
# and we use a dictionary of "'column': 'function'"" pairs

# the result has one column per key, named after the key.

national_parks.groupby('region').agg({
    'visitors': 'mean',
    'unit_name': 'nunique',
    'year': 'min'
})
```

```python
# Write one .agg() call, grouped by region, that reports the median number of visitors, 
# the latest year of record, and the number of distinct states in each region. 
# Why might the median be the more honest of the two averages for this data?

national_parks.groupby('region').agg({
    'visitors': 'median',
    'state': 'nunique',
    'year': 'max'
})
```

```python
# Ranking a grouped result


national_parks.groupby('unit_name')['visitors'].mean().sort_values(ascending=False).head(10)
```

```python
park_stats = national_parks.groupby('unit_name')['visitors'].agg(['count', 'mean', 'max'])

park_stats.sort_values('mean', ascending=False).head(10)
```

```python
# Filter national_parks to the single year '2016', then group by state and use .agg() to report 
# the count and mean of visitors. Rank the result by mean. 
# Compare the top three states with the all-time ranking by state from the earlier exercise, 
# the one that asked for count, mean and max. Did anything move? 


parks_2016 = national_parks[national_parks['year'] == '2016']

stats_2016 = parks_2016.groupby('state')['visitors'].agg(['count', 'mean'])
stats_2016.sort_values('mean', ascending=False).head(10)
```
