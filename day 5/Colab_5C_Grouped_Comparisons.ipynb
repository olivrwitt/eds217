```python
import pandas as pd
import numpy as np

url = 'https://eds-217-essential-python.github.io/data/marine_microplastics.csv'
df = pd.read_csv(url)

df = df.dropna(subset=['Measurement'])
samples = df[df['Unit'] == 'pieces/m3'].copy()
samples['ocean'] = samples['Oceans'].str.replace(' Ocean', '')
positive = samples[samples['Measurement'] > 0].copy()

positive.shape
```

```python
positive.groupby('ocean')['Measurement'].mean().sort_values(ascending = False)
```

```python
positive.groupby('ocean')['Measurement'].median().sort_values(ascending = False)
```

```python
positive['ocean'].value_counts()
```

Southern and arctic oceans have too few samples to count

```python
positive.groupby('ocean')['Measurement'].agg(['count', 'median', 'mean', 'max'])
```

The large maximum for the atlantic ocean definitely skews the mean to be larger.  Therefore I would trust median more.

```python
positive.groupby('ocean').agg({'Measurement': 'mean', 'Sampling Method': 'count', 'Organization': 'count'}) 
```

```python
positive.groupby('Sampling Method')['Measurement'].agg(['count', 'median']).sort_values('median', ascending = False)
```

Largest median: stainless steel spoon (21840)
Smallest median: grab sample (0.0060)


These are separated by six orders of magnitude

If the oceans were not sampled with the same mix of sampling methods, the means won't really be comparable since different methods have different efficiencies and introduce different biases.

```python
by_measurement = positive.groupby(['ocean', 'Sampling Method'])['Sampling Method'].value_counts().sort_values(ascending = False)
by_measurement.reset_index()
```

```python
atlantic = positive[positive['ocean'] == 'Atlantic'].copy()
pacific = positive[positive['ocean'] == 'Pacific'].copy()

print("Atlantic")
print(atlantic.value_counts('Sampling Method'))

print("Pacific")
print(pacific.value_counts('Sampling Method'))
```

```python
neuston_net = positive[positive['Sampling Method'] == 'Neuston net'].copy()

neuston_net.groupby('ocean')['Measurement'].agg(['count', 'median', 'mean'])
```
