```python
a_capella = df[df['genre'] == 'A Capella']
```

```python
plt.figure(figsize =(4,3))
sns.histplot(data=a_capella, x='speechiness')

plt.xlabel('Speechiness')
plt.title('A Capella Speechiness')
plt.show()
```
