```python
import pandas as pd
import matplotlib.pyplot as plt

url = "https://eds-217-essential-python.github.io/data/toolik_weather.csv"
df = pd.read_csv(url)

monthly = df.groupby('Month')
monthly_means = monthly['Daily_AirTemp_Mean_C'].mean()
```

```python
plt.plot(monthly_means)
plt.title("Toolik Monthly Temperatures")
plt.xlabel("Month")
plt.ylabel("Temperature (°C)")
plt.show()
```

```python
plt.plot(monthly_means)
plt.title("Monthly Mean Air Temperature, 1988 to 2018")
plt.xlabel("Month")
plt.ylabel("Temperature (°C)")
plt.show()
```

```python
plt.bar(monthly_means.index, monthly_means)
plt.title("Toolik Monthly Temperatures")
plt.xlabel("Month")
plt.ylabel("Temperature (°C)")
plt.show()
```

```python
coldest = monthly_means.min()
warmest = monthly_means.max()
diff = warmest - coldest
print(f"""The coldest month in Toolik sees an average of about {coldest} degrees celsius.
The warmest month sees an average of about {warmest} degrees celsius.
This is a difference of {diff} degrees celsius.""")
```

```python
monthly_means.to_csv("monthly_means.csv", header=True, float_format="%.2f")
```
