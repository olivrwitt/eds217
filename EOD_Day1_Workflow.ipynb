```python
import pandas as pd
import matplotlib.pyplot as plt

url = "https://eds-217-essential-python.github.io/data/toolik_weather.csv"
df = pd.read_csv(url)
```

```python
df.head()
df.isnull().sum()
```

```python
monthly = df.groupby('Month')
monthly_means = monthly['Daily_AirTemp_Mean_C'].mean()
monthly_means
```

```python
plt.plot(monthly_means)
plt.title("Toolik Monthly Temperatures")
plt.xlabel("Month")
plt.ylabel("Temperature (°C)")
plt.show()
```

```python
monthly_means.to_csv("monthly_means.csv", header=True, float_format="%.2f")
```

```python
windsp_monthly = monthly['Daily_windsp_mean_msec'].mean()
jan_windsp = windsp_monthly[1]
print(f"At Toolik, the average windspeed in January was {jan_windsp:.2f} m/sec.")
```

```python
by_year = df.groupby('Year')
yearly_means = by_year['Daily_AirTemp_Mean_C'].mean()
yearly_means
```

```python
temp_1990 = -8.54
print(f"In 1990, the average temperature in Toolik was {temp_1990:.2f} degrees Celsius.")
```

```python
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
plt.bar(months, monthly_means)
plt.title("Toolik Monthly Temperatures")
plt.xlabel("Month")
plt.ylabel("Temperature (°C)")
plt.show()
```

```python
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
plt.bar(months, monthly_means)
plt.title("Arctic Seasonality, Toolik Lake")
plt.xlabel("Month")
plt.ylabel("Temperature (°C)")
plt.show()
```
