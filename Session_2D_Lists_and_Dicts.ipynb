## 1. Making lists

```python
names = ["Priscilla", "Emalia", "Kailani", "Aaron", "Natasha", "Kelly"]
type(names)
```

## 2. Getting things out of a list

```python
print(names[0])
print(names[5])
print(names[-2])
print(len(names))
print(names[0:5])
```

## 3. Changing a list

```python
names[0] = "Liam"
```

```python
new_students = ["Nina", "Mark"]
names = names + new_students
```

```python
newer_students = ["Ashwin", "Sarah"]
names = names + newer_students
```

```python
names = names[2:]
```

## 4. Looping over a list

```python
for name in names:
    print(f"{name} is great!")
```

## 5. Making dictionaries

```python
names_pets = {'Nina': 1, 'Pirscilla': 7, "Elyse": 5, "Karolina": 0}
```

```python
names_pets["Karla"] = 1
```

## 6. Getting things out of a dictionary

```python
names_pets
```

```python
names_pets["Nina"]
```

```python
names_pets.pop("Karla")
names_pets
```

## 7. Looping over a dictionary

```python
for key, value in names_pets.items():
    print(f"{key} has {value} pets")
    if value > 1:
        print(f"{key} has too many pets!")
```
