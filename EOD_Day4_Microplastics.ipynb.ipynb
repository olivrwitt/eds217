```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

url = 'https://eds-217-essential-python.github.io/data/marine_microplastics.csv'
df = pd.read_csv(url, parse_dates=['Date'], date_format='%m/%d/%Y %I:%M:%S %p')
```

Which columns are object, and which of those should be something else?

Oceans, regions, subregions, sampling method, unit, density range, density class, short reference, long reference, DOI, organization, keywords, accession link, and globalID are all objects.

All of these should be objects

```python
df.isnull().sum()
```

How many columns have gaps? Which are the columns, and how many gaps do they have?

oceans, region, subregion, measurement, and keyword all have missing data.  On rows where oceans is missing, region is also missing.

```python
df = df.dropna(subset=['Measurement'])
df
```

5792 rows were dropped

```python
df['Unit'].value_counts()
```

The "pieces/10 mins: 5,792" row disappeared.  This is because our dropna dropped all 5792 rows with that sampling method. 

```python
# Fill the gaps in Regions and SubRegions with 'Unspecified', one column at a time, 
# assigning each result back. 
# Confirm with .isnull().sum() that only Keywords is still missing anything

df['Regions'] = df['Regions'].fillna('Unspecified')
df['SubRegions'] = df['SubRegions'].fillna('Unspecified')
df.isnull().sum()
```

```python
# Check for duplicated rows. 
# Do it twice: once bare, and once with subset= naming the columns that define a 
# single observation, which here are Latitude, Longitude, Date, Measurement and Unit. 
# Report both numbers and say which one you believe, and why the two differ so much.

print(df.duplicated().sum())
print(df.duplicated(subset=['Latitude', 'Longitude', 'Date', 'Measurement', 'Unit']).sum())
```

df.duplicated said there were zero duplicates. Adding a subset changed the number of duplicates rows to 539.  The difference is probably because there are no rows where ALL columns are the same, but there are many rows where just the 5 subsetted columns are the same.  Which value is "correct" depends on the purpose.  It is likely that some of the subsetted "duplicates" could not actually be duplicates if multiple measurements are collected from the same site on the same day.
