```python
import pandas as pd

url = 'https://eds-217-essential-python.github.io/data/marine_microplastics.csv'
plastics = pd.read_csv(url)
plastics.shape
```

```python
# this is a function:
def celsius_to_fahrenheit(celsius):
    fahrenheit = (celsius * 9 / 5) + 32
    return fahrenheit

# def announces that a definition is coming
# celsius_to_fahrenheit is the name
# (celsius) is the parameter
# everything indented under the def line is the body

# use it with a function call:
celsius_to_fahrenheit(25)
```

```python
# A parameter can have a default, which is used when the caller does not supply that argument:

def convert_depth(depth, units='meters'):
    """Convert a depth in metres to metres, feet, or fathoms."""
    if units == 'meters':
        return depth
    elif units == 'feet':
        return depth * 3.28084
    elif units == 'fathoms':
        return depth * 0.546807
    else:
        return None
```

```python
print(convert_depth(100))
print(convert_depth(100, 'feet'))
```

```python
# Keyword arguments:
#  When you call a function, you can name the arguments instead of relying on their order:

print(convert_depth(depth=100, units='fathoms'))
print(convert_depth(units='fathoms', depth=100))
```

```python
# Write a function classify_latitude that takes a latitude and returns 'tropical' 
#  if it is between -23.5 and 23.5, 'polar' if it is below -66.5 or above 66.5,
#   and 'temperate' otherwise. 
# Test it on 0, 45, and -70.

def classify_latitude(latitude):
    if latitude > -23.5 and latitude < 23.5:
        climate = "Tropical"
    elif latitude > 66.5 or latitude < -66.5:
        climate = "Polar"
    else:
        climate = "Temperate"
    return climate


print(classify_latitude(0))
print(classify_latitude(45))
print(classify_latitude(-70))
```
