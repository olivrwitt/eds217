```python
# Filter first and then rank
goleta[goleta['parameter'] == 'pm25'].sort_values('value', ascending=False).head(5)[
    ['datetimeLocal', 'value', 'unit']
]
# Why?  Particulates and ozone are measured in different units, so 
# the values in the 'values' column arent comparable unless 
# filtered to only one pollutant.
```

```python
# .idxmax gives the index label of the largest value in a column
# i.e. "index max"
worst_hour = pm25['value'].idxmax()
pm25.loc[worst_hour, 'datetimeLocal']
```

```python
# idxmin = index minimum - gives index label of smallest
# value in column
cleanest_hour = pm25['value'].idxmin()
print(cleanest_hour)
print(pm25.loc[cleanest_hour, 'datetimeLocal'])
print(pm25.loc[cleanest_hour, 'value'])
```

```python
# .copy()
# creates a new table so that you can keep and modify a filtered result
# this way, you are no longer tied to the original table

pm10 = goleta[goleta['parameter'] == 'pm10'].copy()
```

```python
worst_ten = pm10.sort_values('value', ascending=False).head(10)
worst_ten[['datetimeLocal', 'value', 'unit']]
print("Worst pm10 hour")
```
