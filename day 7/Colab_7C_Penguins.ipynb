```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

url = 'https://eds-217-essential-python.github.io/data/penguins.csv'
penguins = pd.read_csv(url)
```

```python
penguins.shape
```

```python
penguins.head()
```

```python
penguins.isnull().sum()
```

```python
birds = penguins.dropna()
birds.shape
```

```python
plt.figure(figsize=(7, 5))

sns.histplot(data=birds, x='flipper_length_mm')

plt.xlabel('Flipper Length')
plt.title('Flipper Length of Penguins')
plt.show()
```

```python
plt.figure(figsize=(7, 5))

sns.histplot(data=birds, x='flipper_length_mm', hue='species')

plt.xlabel('Flipper Length')
plt.title('Flipper Length by Species')
plt.show()
```

```python
plt.figure(figsize=(7, 5))

sns.histplot(data=birds, x='bill_length_mm', hue='species')

plt.xlabel('Bill Length')
plt.title('Bill Length by Species')
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(data=birds, x='body_mass_g', y='flipper_length_mm', hue='species')

plt.xlabel('Body Mass (g)')
plt.ylabel('Flipper Length (mm)')
plt.title('Penguin Body Mass vs Flipper Length, by Species')
plt.show()
```
