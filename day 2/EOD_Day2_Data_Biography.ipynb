# Part 4: The numbers

### 1 
What is the mean daily air temperature across the whole record? The minimum? The maximum?

```python
print(f"Mean: {df['Daily_AirTemp_Mean_C'].mean()}")
print(f"Max: {df['Daily_AirTemp_Mean_C'].max()}")
print(f"Min: {df['Daily_AirTemp_Mean_C'].min()}")
```

The mean daily air temperature across the whole record is -8.2 degrees celsius.  The maximum was 20.7 degrees celsius, while the minimum was -55.6 degrees celsius.

### 2
Are the mean, the minimum and the maximum plausible for a site above the Arctic Circle? Say why or why not.

These are plausible.  The maximum is close to 70 degrees F, which seems a little unlikely but not impossible.  The minimum and average are quite plausible for the arctic.

### 3
Pick one other numeric column and report its range. Does anything about it look wrong?

```python
print(f"Mean: {df['Daily_Precip_Total_mm'].mean()}")
print(f"Max: {df['Daily_Precip_Total_mm'].max()}")
print(f"Min: {df['Daily_Precip_Total_mm'].min()}")
```

The maximum is 64.5mm, the minimum is 0mm, and the mean is 0.95mm.  This seems plausible, since the max could be a severe snowstorm and the minimum could be a day with no precipitation.
