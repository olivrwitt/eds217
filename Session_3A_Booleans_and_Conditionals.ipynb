## 2. The comparison operators

```python
print(temperature > 0)
print(temperature < 0)
print(temperature >= 0)
print(temperature <= 0)
print(temperature == 8.4)
# == is comparison, and = is assignment!
print(temperature != 8.4)
```

```python
print(toolik.loc[0, 'Station'])
print(toolik.loc[0, 'Station'] == 'TLKMAIN')
```

## 3. Asking two questions at once

```python
temperature = 8.4
precipitation = 0.0
print(temperature > 0 and precipitation == 0.0)
# true when both sides are true
print(temperature == 8.4 and precipitation > 0)
print(temperature == 8.4 or precipitation > 10)
# true when one side is true
print(not temperature > 0)
# flips the answer
```

## 4. if / elif / else

```python
temperature = 8.4
if temperature > 0:
    print ("Above Freezing")

print("Hello")
```

```python
temperature = 8.4
if temperature < 0:
    print ("Below Freezing")

print("Hello")
```

```python
if temperature > 0:
    print ("Above Freezing")
else:
        print("Below Freezing")
```

```python
temperature = 12.0

if temperature > 10:
    category = 'mild'
elif temperature > 0:
    category = 'cool'
else:
    category = 'cold'

print(category)
```

## 5. Asking a whole column

```python
for value in toolik.loc[[0, 1, 2], 'Daily_AirTemp_Mean_C']:
    print(value)
    if value > 7:
        print(f"{value} is warm")
    elif value > 0:
        print(f"{value} is mild")
    else:
        print(f"{value} is cold")
```

```python
# a whole column at once

toolik['Daily_AirTemp_Mean_C'] > 0
```

```python
# Rules for logical operators and Data Frames / Series
# 1. Wrap every conditional in parantheses
# 2. When dong "ands" and "ors", we switch to "&" and "|"

toolik['Daily_AirTemp_Mean_C'] > 0
toolik['Daily_Precip_Total_mm'] == 0
```

```python
warm_and_dry = (toolik['Daily_AirTemp_Mean_C'] > 0) & (toolik['Daily_Precip_Total_mm'] == 0)
print(warm_and_dry)
```

## 6. Counting the Trues

```python
above_zero = toolik['Daily_AirTemp_Mean_C'] > 0
above_zero.sum()
```

```python
print(above_zero.sum() / toolik.shape[0])
print(above_zero.sum() / len(toolik))
print(sum(above_zero) / len(toolik))
```
